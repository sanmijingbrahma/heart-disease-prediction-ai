# Data Cleaning

## Objective
- Tranform raw cleaveland heart disease dataset to clean datset that is ready for EDA and model training

## Import libraries and load dataset

```python
import pandas as pd

df = pd.read_csv("../data/raw/cleveland.csv")

df.head()
```

## Missing Values

```python
(df == "?").sum()
```

### Replace with null

```python
df.replace("?",pd.NA,inplace=True)
```

### Convert them to numerical Datatype

```python
df["ca"] = pd.to_numeric(df["ca"])
df["thal"] = pd.to_numeric(df["thal"])

df.info()
```

### Handle Missing values
- Median for "ca", as it is ordinal/discrete
- Mode for "thal", as it is categorical

```python
df["ca"].fillna(df["ca"].median(),inplace=True)

df["thal"].fillna(df["thal"].mode()[0],inplace=True)

df.info()
```

## Binary Target
- As our probelm statement is only about "Predict whetehr a patient has a heart disease or not"

```python
df["target"] = (df["num"]>0).astype(int)
```

```python
df["target"].value_counts()
```

### Drop Old target

```python
df.drop(columns="num",inplace=True)
```

## Verification

```python
df.info()
```

```python
df.describe()
```

```python
df.isna().sum()
```

```python
df.head()
```

## Save Dataset

```python
df.to_csv("../data/processed/cleveland_clean.csv", index=False)
```


## Cleaning Summary

- Replaced hidden missing values ("?") with NaN.
- Converted `ca` and `thal` to numeric data types.
- Imputed missing values using the median (`ca`) and mode (`thal`).
- Created a binary target (`target`) where:
  - 0 = No Heart Disease
  - 1 = Heart Disease
- Removed the original multiclass target (`num`).
- Saved the cleaned dataset as `data/processed/cleveland_clean.csv`.
