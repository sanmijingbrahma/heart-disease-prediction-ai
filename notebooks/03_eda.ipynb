# Bivariate Analysis

Understanding whcih are related to which

## Age vs Target

```python
plt.figure(figsize=(7,5))
sns.boxplot(
    data=df,
    x="target",
    y="age"
)
plt.title("Age vs Heart Disease")
plt.xlabel("Target")
plt.ylabel("Age")
plt.show()
```

Here we can see that median of individual with heart disease higher, so according to dataset
we can say that individual with heart diseases tend to be older

## Cholestoral vs Target

```python
plt.figure(figsize=(7,5))
sns.boxplot(
    data=df,
    x="target",
    y="chol"
)
plt.title("Cholestoral vs Heart Disease")
plt.xlabel("Target")
plt.ylabel("Cholestoral(mg/dl)")
plt.show()
```

Here we can also say that people with higher colestoral tend to have comparivley more 
heart diseases according to dataset

## Resting Blood Rate vs Heart Disease

```python
plt.figure(figsize=(7,5))
sns.boxplot(
    data=df,
    x="target",
    y="trestbps"
)
plt.title("Resting Bolood Rate vs Heart Disease")
plt.xlabel("Target")
plt.ylabel("Resting Blood Rate (mmHg)")
plt.show()
```

As per the plot we see shat median of resting blood pressure are almost similar, but the only differene is that people with heart diseases have more spread of resting blood pressure compared to those without heart disease

## Maximum Heart Rate vs Heart Disease

```python
plt.figure(figsize=(7,5))
sns.boxplot(
    data=df,
    x="target",
    y="thalach"
)
plt.title("Maximum Heart Rate vs Heart Disease")
plt.xlabel("Target")
plt.ylabel("Maximum Heart Rate(bps)")
plt.show()
```

Here as per box plot we have seen that people without heart disease have higher maximum heart rate when compared to those with heart diseases, so in a way we can say that comparitively those without any heart diseases tend to achieve higher heart rates.

## Chest Pain vs Heart Disease

```python
plt.figure(figsize=(7,5))
sns.countplot(
    data=df,
    x="cp",
    hue="target"
)
plt.title("Chest Pain vs Heart Disease")
plt.xlabel("Chest Pain")
plt.ylabel("Frequency")
plt.legend(title="Heart Disease",labels=["No","Yes"])
plt.show()
```

- Type 1 Chest Pain
    - This type fo chest pain is more common to those without any heart diseases
- Type 2 Chest Pain
    - This type fo chest pain is more common to those without any heart diseases
- Type 3 Chest Pain
    - This type fo chest pain is more common to those without any heart diseases
- Type 4 Chest Pain
    - This type fo chest pain is more common to those with heart diseases

## Exercise Induced Angina vs Heart Disease

```python
plt.figure(figsize=(7,5))
sns.countplot(
    data=df,
    x="exang",
    hue="target"
    
)
plt.title("Exercise Induced Angina vs Heart Disease")
plt.xlabel("Exercise Induced Angina")
plt.ylabel("Frequency")
plt.legend(title="Heart Disease", labels=["No", "Yes"])
plt.show()
```

Here we can notice that exercise induced Angina are more likely to have heart disease, which suggest a possible association between exercise induced angina wtih heart disease.

# Correlation Analysis

```python
## Only for Numerical columns
corr = df.corr(numeric_only=True)
```

```python
plt.figure(figsize=(14,14))
sns.heatmap(corr,annot=True,cmap="coolwarm",fmt=".2f")
plt.title("Correlation Heamap")
plt.show()
```

- target is most correlated with "thal"(Results of Nuclear stress test) of 0.53
- Slope and oldpeak are most highly correlated with 0.58

- As per the correlation heatmap none of the features have very high correlation therefor there is not
eveidencd of severe multicollinearity among the features.
